groupby

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "category" : ['a','b','a','b','a','b'],
    "store" : ['s1','s2','s1','s2','s1','s2',],
    'sales': [100,200,240,400,120,230],
    'quantity' : [10,15,12,10,13,14]
})
```

```python
df.groupby('store')['sales'].sum()

```

groupby in multiple columns

```python
df.groupby(['category','store'])[['sales','quantity']].sum()
```

Aggregation

```python
#function = mean , median , mode , min , max,std
df['sales'].max()
```

```python
df['sales'].agg(['mean','median', "max"])
```

Pivot table

```python
dp = pd.DataFrame({
    "date" : pd.date_range('2026-08-01',periods = 8),
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West"],
        "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Laptop", "Phone", "Tablet"],
        "Salesperson": ["John", "Sarah", "Mike", "Emma", "John", "Sarah", "Mike", "Emma"],
        "Units": np.random.randint(100,200,8),
        "Revenue":np.random.randint(100,1000,8) 
})
dp         
```

```python
pd.pivot_table(dp,values='Revenue', index= "Region", columns= "Product",aggfunc = "median")
```

Cross Table

```python
pd.crosstab(dp["Region"], dp["Product"])
#used for counting purpose
```
