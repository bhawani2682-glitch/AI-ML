 Craeting a Dataframe

```python
import numpy as np
import pandas as pd
```

```python
dict = {
    "name":["bhawani","bhuvan","divyansh","dhruv", "dipanshu"],
        "age":[20,20,21,20,20],
        "height":[5.11,5.8,5.9,5.8,5.9],
        "weight":[65, 67,62,63,70],
        "grade":["A","A", "A","A","A"],
}
df = pd.DataFrame(dict)
print(df)
```

Selection and Indexing of columns

```python
# select a column
df[['age','name']]
# creating a new columns
df['address']= ["home","home","home","home","home"]
#remove a column
df.drop('address', axis = 1, inplace = True)
df
```

Selection and Indexing of rows

```python
# select a row
df.loc[[0,1]]
#selecting a specific position
df.loc[[0,1]][['name','age']]
# creating a new rows
df.loc[4] = ['dipanshu',21,5.78,64,'A']
#remove a row
df.drop(5,axis =0,inplace= True)
df

```

conditinal selection


```python
df[df["age"]>20]
df[(df["age"]==20 )&(df["weight"] <69)]
```
