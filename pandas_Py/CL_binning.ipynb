Equal width binning

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "age": [15,20,25,30,35,40,45,50,55,60,65,70]
})
df["age_bin"] = pd.cut(
    df["age"],
    bins = 3,
    labels = ["very young","young", "very senior"]
)
print(df)
```

Equal - frequency Binning


```python
df = pd.DataFrame({
    "age": [15,20,25,30,35,40,45,50,55,60,65,70]
})
df["age_bin"] = pd.qcut(
    df["age"],
    q=3,
     labels = ["very young","young", "very senior"]
)
print(df)
```

Custom Binning


```python
df = pd.DataFrame({
    "age": [15,20,25,30,35,40,45,50,55,60,65,70]
})
bins = [20,35,50,60]
labels = ["very young","young", "very senior"]
df["age_group"] = pd.cut(df["age"], bins = bins, labels = labels)
print(df)
```
