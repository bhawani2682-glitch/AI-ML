Advance opertion using big data set

```python
import numpy as np
from matplotlib.pyplot import *
```

```python
sales_data = np.array([
    ["1",12000,15000,19000,25000,50000],
    ["2",10000,18000,23000,35000,39000],
    ["3",15000,15500,13000,20000,50000],
    ["4",12000,19000,21000,15000,45000],
    ["5",11000,25000,33000,25000,40000]
])
```

```python
print(sales_data.shape)
print(sales_data[:, 1:])
```

total sales per year

```python
print(np.sum(sales_data[:, 1:],axis=0)) #column wise
print(np.sum(sales_data[:, 1:],axis=1)) #row wise
```

minimun and maximun sale

```python
print(np.min(sales_data[:, 1:],axis=1))
print(np.max(sales_data[:, 1:],axis=1))
print(np.average(sales_data[:, 1:],axis=1))
```

```python
cumsum = np.cumsum(sales_data[:, 1:],axis = 1)
f = figure(figsize= (5,3))  
h = hist(cumsum , bins = 6) 
show() 
```

vector opertion

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([9,8,7,6,5])
print(vector1 + vector2)
print(vector1 * vector2)
#Dot product
print(np.dot(vector1,vector2))
```

vectorising an array

```python
restaurant = np.array(["pizza_hut", "dominos" , "haldiram", "pizza18" , "kfc"])
value = np.vectorize(str.upper)
print(value(restaurant))
```

broadcast

```python
monthly_salary = sales_data[:, 1:] /12
print(monthly_salary)
```
