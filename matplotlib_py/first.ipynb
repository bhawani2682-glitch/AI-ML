```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
a = np.arange(0,15)
print(a)
b = a-5
print(b)

```

```python
plt.plot(a,b)
plt.title("Axis")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

Subplots

```python
plt.subplot(2,2,1)
plt.plot(b,a)
plt.subplot(2,2,3)
plt.plot(.5*a,b)
plt.subplot(2,2,2)
plt.subplot(2,2,4)
```

Object oriented way

```python
fig = plt.Figure()
axis1 =fig.add_axes([0.1,0.1,1,1])
axis1.plot(a,b)
axis2 =fig.add_axes([0.1,0.1,1,1])
axis2.plot(a,b)
axis1
```

Types of plot

```python
plt.scatter(a,b)   #useful for regression
```

```python
plt.hist(a)
```

```python
plt.boxplot(a)
```

Saving Plots

```python
fig = plt.Figure()
ax =fig.add_axes([0.1,0.1,.8,.8])
ax.plot(a,b)
plt.savefig("new.png")
```
