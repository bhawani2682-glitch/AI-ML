Categorical Data Plots

```python
#Now let's discuss using seaborn to plot categorical data:
# 1.boxplot
# 2.violinplot
# 3.stripplot
# 4.swarmplot
# 5.barplot
# 6.conutplot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df=sns.load_dataset("tips")
df
```

```python
plt.subplot(1,2,1)
sns.countplot(x =df['sex'], hue =df['smoker'] )
```

Barplot

```python
sns.barplot(x = df['sex'], y = df["total_bill"], estimator=np.sum)
```

Boxplot

```python
sns.boxplot(x = "total_bill", y = 'sex', data =df)
```

stripplot & swarmplot

```python
sns.stripplot(x = 'total_bill', y = 'sex', data = df)
```

```python
sns.swarmplot(x = 'total_bill', y = 'sex', data = df)
```

violinplot

```python
sns.violinplot(x = 'total_bill', y = 'sex', data = df)
```

countplot

```python
sns.countplot(df)
```
