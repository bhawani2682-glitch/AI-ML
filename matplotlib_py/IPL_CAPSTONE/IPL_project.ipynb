```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
ipl = pd.read_csv("Book_ipl22_ver_33.csv")
ipl.head()
```

Basic questions from dataset

```python
#Which team won the most match
match_won =ipl["match_winner"].value_counts()
sns.barplot(y = match_won.index , x = match_won.values,palette= "rainbow")
```

```python
#toss decision Trends
toss =ipl["toss_decision"]
sns.countplot(toss,palette= "rainbow")
```

```python
# toss wins vs match wins
count = ipl[ipl["toss_winner"] == ipl["match_winner"]]["match_id"].count()
percentange = (count * 100)/74
percentange.round(2)
```

```python
#how do teams wins.(Runs vs wickets)
ipl['won_by'].value_counts()
```

Key players performance

```python
#most player of the match awards
counts =ipl['player_of_the_match'].value_counts().head(10)
```

```python
#top scorers
high = ipl.groupby('top_scorer')['highscore'].sum().sort_values(ascending= False).head(5)
high.plot(kind = 'barh')
```

```python
#top 10 best bolling figure
ipl["highest_wickets"] = ipl["best_bowling_figure"].apply(lambda x : x.split("--")[0])
ipl['highest_wickets'] = ipl['highest_wickets'].astype(int)
top_bowler = ipl.groupby("best_bowling")["highest_wickets"].sum()
gg =top_bowler.sort_values(ascending= False).head(5)
gg.plot(kind = 'barh')
```

Some custom questions 

```python
#who won the highest margin by runs
ipl[ipl["won_by"] == "Runs"].sort_values(by ="margin" , ascending = False)[["match_winner", "margin"]].head(1)
```

```python
#which player have the highest individual score
ipl[ipl['highscore'] == ipl['highscore'].max()].head(5)[["top_scorer" , 'highscore']]
```

```python
#which player hade the best bowling figure
ipl[ipl['highest_wickets']==ipl['highest_wickets'].max()].tail(1)[['best_bowling', "best_bowling_figure"]]
```
