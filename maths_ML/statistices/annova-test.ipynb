```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset("titanic")
df
```

```python
df = df["age"].fillna(df['age'].mean(),inplace= True)
```

```python
df.info()
```

```python
df['pclass'].unique()
```

```python
class1 =df[df['pclass'] == 1]['age']
class2 =df[df['pclass'] == 2]['age']
class3 =df[df['pclass'] == 3]['age']
```

```python
f_stats, p_value = f_oneway(class1,class2,class3)
```

```python
f_stats
```

```python
p_value
alpha = .05
```

```python
if p_value < alpha:
    print("i will reject")
else:
    print("i will accept")  
```
